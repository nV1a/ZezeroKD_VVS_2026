```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('titanic.csv')
n = int(input("Введите колво строк: "))
display(df.head(n))
display(df.tail(5))
print(f"Строк: {df.shape[0]}, Столбцов: {df.shape[1]}")
print("Столбцы:", df.columns.tolist())
print("Типы данных:", df.dtypes)
num_f = ['Age', 'Fare', 'SibSp', 'Parch']
cat_f = ['Pclass', 'Sex', 'Embarked', 'Survived', 'Cabin', 'Name', 'Ticket']
display(df[num_f].describe())
print('Пассажиры по классам:', df['Pclass'].value_counts())
print('Уникальных портов посадки:', df['Embarked'].nunique())
print('Пропуски по столбцам:', df.isnull().sum())
print('Процент пропусков:', df.isnull().mean() * 100)
max_missing_col = df.isnull().sum().idxmax()
print(f'Больше всего пропусков в столбце: {max_missing_col}')
age_before = {
    'Среднее': df['Age'].mean(),
    'Медиана': df['Age'].median(),
    'Минимум': df['Age'].min(),
    'Максимум': df['Age'].max(),
    'Ст. отклонение': df['Age'].std(),
    'Колво': df['Age'].count()
}

age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)
age_after = {
    'Среднее': df['Age'].mean(),
    'Медиана': df['Age'].median(),
    'Минимум': df['Age'].min(),
    'Максимум': df['Age'].max(),
    'Ст. отклонение': df['Age'].std(),
    'Колво': df['Age'].count()
}

comparison_df = pd.DataFrame([age_before, age_after], index=['До обработки', 'После обработки'])
display(comparison_df)
shape_before = df.shape
df = df.dropna(subset=['Embarked'])
if df['Cabin'].isnull().mean() > 0.5:
  df = df.drop(columns=['Cabin'])
print('Остаток пропусков:', df.isnull().sum())
print(f'Размер до: {shape_before}, после: {df.shape}')
survived_arr = df['Survived'].to_numpy()
sex_arr = df['Sex'].to_numpy()
age_arr = df['Age'].to_numpy()
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
fare_arr = df['Fare'].to_numpy()
survival_rate = np.mean(survived_arr) * 100
print(f'Общий процент выживших: {survival_rate:.2f}%')
male_survived = np.mean(survived_arr[sex_arr == 'male']) * 100
female_survived = np.mean(survived_arr[sex_arr == 'female']) * 100
male_age = np.mean(age_arr[sex_arr == 'male'])
female_age = np.mean(age_arr[sex_arr == 'female'])
male_surv_age = np.mean(age_arr[(sex_arr == 'male') & (survived_arr == 1)])
male_dead_age = np.mean(age_arr[(sex_arr == 'male') & (survived_arr == 0)])
female_surv_age = np.mean(age_arr[(sex_arr == 'female') & (survived_arr == 1)])
female_dead_age = np.mean(age_arr[(sex_arr == 'female') & (survived_arr == 0)])
print(f'Билет - мин: {np.min(fare_arr):.2f}, макс: {np.max(fare_arr):.2f}, среднее: {np.mean(fare_arr):.2f}')
print(f'Ст. отклонение возраста: {np.std(age_arr):.2f}')
numpy_stats = pd.DataFrame({
    'Процент выживших': [male_survived, female_survived],
    'Средний возраст': [male_age, female_age],
    'Ср. возраст выживших': [male_surv_age, female_surv_age],
    'Ср. возраст погибших': [male_dead_age, female_dead_age]
}, index=['Мужчины', 'Женщины'])
display(numpy_stats)

task1 = df[(df['Sex'] == 'male') & (df['Age'] > 30) & (df['Pclass'] == 1)]
print('Task1', len(task1))
display(task1.head(10))

task2 = df[((df['Age'] < 18) | (df['Sex'] == 'female')) & (df['Survived'] == 1)]
print('Task2', len(task2))

mean_fare_3 = df[df['Pclass'] == 3]['Fare'].mean()
task3 = df[(df['Pclass'] == 3) & (df['Fare'] > mean_fare_3)]
print('Task3', len(task3))

task4 = df[(df['Age'] >= 20) & (df['Age'] <= 40) & (df['Embarked'] == 'S')]
print('Task4', len(task4))


group_class_sex = df.groupby(['Pclass', 'Sex']).agg(
    Количество=('PassengerId', 'count'),
    Средний_возраст=('Age', 'mean'),
    Доля_выживших_процент=('Survived', lambda x: x.mean() * 100),
    Средняя_цена_билета=('Fare', 'mean')
).reset_index()
display(group_class_sex)

group_embarked = df.groupby('Embarked').agg(
    Количество=('PassengerId', 'count'),
    Доля_выживших_процент=('Survived', lambda x: x.mean() * 100)
).reset_index()
display(group_embarked)

best_class = df.groupby('Pclass')['Survived'].mean().idxmax()
print(f'Класс с наиб долей выживших: {best_class}')

df.to_csv('titanic_processed.csv', index=False)
group_class_sex.to_csv('titanic_statistics.csv', index=False)
df_test = pd.read_csv('titanic_processed.csv')
print('Успех', len(df_test))
```

Пропуски: Максимальное количество в столбцах Cabin (327 шт. / 78.2%) и Age (86 шт. / 20.5%)

Влияние обработки: Столбец Cabin удален из-за обилия пропусков. Заполнение пропусков возраста медианой (27.0) снизило средний возраст с 30.27 до 29.60 и уменьшило стандартное отклонение с 14.18 до 12.70 (данные стали кучнее)

Выживаемость по полу: В данном датасете выжило 100% женщин и 0% мужчин.

Выживаемость по классу: Максимальная доля выживших зафиксирована в 1 классе.

Полезные инструменты: В Pandas - fillna() для безопасного заполнения пропусков, логические маски (&, |) для фильтрации и groupby() для агрегации; в NumPy - np.mean() и np.std() для быстрых расчетов.
